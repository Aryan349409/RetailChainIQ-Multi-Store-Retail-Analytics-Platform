```python
df.drop_duplicates(inplace=True)
```

```python
df.dtypes
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.replace(".", "_", regex=False)
      .str.replace(" ", "_")
)
```

```python
df.columns.tolist()
```

```python
columns_to_drop = ["记录数", "Year", "weeknum", "Market2"]

df.drop(columns=[col for col in columns_to_drop if col in df.columns], inplace=True)
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Ship_Date"] = pd.to_datetime(df["Ship_Date"])
```

```python
df.info()
```

```python
df["Category"].unique()
```

```python
df["Sub_Category"].unique()
```

```python
df["Region"].unique()
```

```python
df["Market"].unique()
```

```python
df["Order_Year"] = df["Order_Date"].dt.year
```

```python
df["Quarter"] = df["Order_Date"].dt.quarter
```

```python
df["Month_Number"] = df["Order_Date"].dt.month
```

```python
df["Month_Name"] = df["Order_Date"].dt.month_name()
```

```python
df["Week_Number"] = df["Order_Date"].dt.isocalendar().week
```

```python
df["Day_Name"] = df["Order_Date"].dt.day_name()
```

```python
df["Weekend"] = df["Day_Name"].isin(["Saturday", "Sunday"])
```

```python
df["Delivery_Days"] = (df["Ship_Date"] - df["Order_Date"]).dt.days
```

```python
df["Profit_Margin"] = np.where(
    df["Sales"] != 0,
    (df["Profit"] / df["Sales"]) * 100,
    0
)
```

```python
df["Discount_Category"] = pd.cut(
    df["Discount"],
    bins=[-0.01, 0, 0.10, 0.25, 1],
    labels=["No Discount", "Low", "Medium", "High"]
)
```

```python
df["Sales_Band"] = pd.cut(
    df["Sales"],
    bins=[0, 100, 500, 1000, 5000, df["Sales"].max()],
    labels=["Very Low", "Low", "Medium", "High", "Very High"],
    include_lowest=True
)
```

```python
df["Order_Size"] = pd.cut(
    df["Quantity"],
    bins=[0, 2, 5, df["Quantity"].max()],
    labels=["Small", "Medium", "Large"],
    include_lowest=True
)
```

```python
df["Profit_Category"] = pd.cut(
    df["Profit"],
    bins=[df["Profit"].min()-1, 0, 100, 500, df["Profit"].max()],
    labels=["Loss", "Low Profit", "Medium Profit", "High Profit"]
)
```

```python
season_map = {
    "January":"Winter",
    "February":"Winter",
    "March":"Spring",
    "April":"Spring",
    "May":"Spring",
    "June":"Summer",
    "July":"Summer",
    "August":"Summer",
    "September":"Autumn",
    "October":"Autumn",
    "November":"Autumn",
    "December":"Winter"
}

df["Season"] = df["Month_Name"].map(season_map)
```

```python
customer_orders = (
    df.groupby("Customer_ID")
      .size()
      .reset_index(name="Order_Count")
)
```

```python
df = df.merge(customer_orders, on="Customer_ID", how="left")
```

```python
df["Customer_Type"] = np.where(
    df["Order_Count"] > 1,
    "Returning",
    "New"
)
```
